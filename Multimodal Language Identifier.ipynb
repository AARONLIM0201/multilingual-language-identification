```python
def crawl_wikipedia(seed_url, base_url, max_articles=10):
    headers = {"User-Agent": "Mozilla/5.0 (FYP crawler)"}
    visited = set()
    article_urls = set()
    queue = deque([seed_url])

    def normalize_url(href: str) -> str:
        # Support /wiki/, /zh-hans/, /zh-my/ link patterns
        if href.startswith("/wiki/"):
            return base_url + href
        if href.startswith("/zh-hans/") or href.startswith("/zh-my/"):
            return base_url + href
        # Ignore other namespaces and external links
        return ""

    while queue and len(article_urls) < max_articles:
        url = queue.popleft()
        if url in visited:
            continue
        visited.add(url)

        try:
            r = requests.get(url, headers=headers, timeout=10)
            soup = BeautifulSoup(r.content, "html.parser")

            # Collect more internal links to crawl
            for a in soup.select("a[href]"):
                href = a.get("href") or ""
                if ":" in href:
                    # skip special pages like /wiki/Special:... or interwiki
                    continue
                full_url = normalize_url(href)
                if not full_url:
                    continue
                article_urls.add(full_url)
                if len(article_urls) >= max_articles:
                    break

            time.sleep(0.3)

        except Exception as e:
            print("Crawler error:", e)

    return list(article_urls)
```

```python
# Tatoeba CN loader (optional)
import csv
import re
import random

TATOEBA_SENT_PATH = "sentences.csv"  # download from https://downloads.tatoeba.org/exports/sentences.csv
TATOEBA_MAX_ROWS = 200000             # cap for speed; adjust if needed


def load_tatoeba_cn(path=TATOEBA_SENT_PATH, max_rows=TATOEBA_MAX_ROWS):
    """Load Mandarin sentences (cmn/zh), convert to Simplified if enabled, drop ASCII if configured."""
    rows = []
    langs = {"cmn", "zh"}
    with open(path, encoding="utf-8") as f:
        reader = csv.reader(f, delimiter="\t")
        for i, row in enumerate(reader):
            if len(row) < 3:
                continue
            _, lang, text = row[0], row[1], row[2].strip()
            if lang not in langs:
                continue
            if CN_ENSURE_SIMPLIFIED and cc:
                text = cc.convert(text)
            if CN_FILTER_ASCII and re.search(r"[A-Za-z]", text):
                continue
            if text:
                rows.append(text)
            if i >= max_rows:
                break
    return rows


def chunk_cn_texts(texts, min_len=20, max_sentences=100):
    """Split CN texts, keep >= min_len, chunk to ~18 chars and merge short pieces."""
    sentences = []
    for text in texts:
        sents = re.split(r'[。！？]', text)
        for sent in sents:
            sent = sent.strip()
            if not sent:
                continue
            if len(sent) >= min_len:
                sentences.append(sent)
            if len(sentences) >= max_sentences:
                break
        if len(sentences) >= max_sentences:
            break

    random.shuffle(sentences)
    sentences = sentences[:max_sentences]

    chunk_size = 18
    min_chunk = 8
    tmp = []
    for sent in sentences:
        s = sent
        while s:
            chunk = s[:chunk_size]
            tmp.append(chunk)
            s = s[chunk_size:]

    grouped = []
    i = 0
    while i < len(tmp):
        curr = tmp[i]
        if len(curr) < min_chunk and i + 1 < len(tmp):
            merged = curr + tmp[i + 1]
            grouped.append(merged)
            i += 2
        else:
            grouped.append(curr)
            i += 1

    return grouped[:max_sentences]
```

```python
def extract_sentences(urls, lang, min_len=20, max_sentences=100):
    import re
    headers = {"User-Agent": "Mozilla/5.0"}
    sentences = []

    for url in urls:
        if len(sentences) >= max_sentences:
            break
        try:
            r = requests.get(url, headers=headers, timeout=10)
            soup = BeautifulSoup(r.content, "html.parser")

            for p in soup.find_all("p"):
                text = p.get_text().strip()
                if lang == "CN":
                    # Ensure simplified if enabled
                    if CN_ENSURE_SIMPLIFIED and cc:
                        text = cc.convert(text)
                    # Split using Chinese sentence delimiters
                    sents = re.split(r'[。！？]', text)
                else:
                    sents = re.split(r'[.!?]', text)
                for sent in sents:
                    sent = sent.strip()
                    if not sent:
                        continue
                    # Drop English letters for CN if enabled
                    if lang == "CN" and CN_FILTER_ASCII and re.search(r"[A-Za-z]", sent):
                        continue
                    if len(sent) >= min_len:
                        sentences.append(sent)
                    if len(sentences) >= max_sentences:
                        break

            time.sleep(0.2)
        except Exception:
            continue

    random.shuffle(sentences)
    sentences = sentences[:max_sentences]

    grouped = []
    if lang == "CN":
        chunk_size = 18           # characters per chunk
        min_chunk = 8             # merge if shorter than this
        tmp = []
        for sent in sentences:
            s = sent
            while s:
                chunk = s[:chunk_size]
                tmp.append(chunk)
                s = s[chunk_size:]
        # merge short chunks with the next one to avoid very short tiles
        i = 0
        while i < len(tmp):
            curr = tmp[i]
            if len(curr) < min_chunk and i + 1 < len(tmp):
                merged = curr + tmp[i + 1]
                grouped.append(merged)
                i += 2
            else:
                grouped.append(curr)
                i += 1
    else:
        max_words = 12            # words per chunk
        min_words = 3             # merge if shorter than this
        tmp = []
        for sent in sentences:
            words = sent.split()
            for i in range(0, len(words), max_words):
                chunk = words[i:i + max_words]
                if chunk:
                    tmp.append(chunk)
        # merge short chunks with the next chunk
        i = 0
        while i < len(tmp):
            curr = tmp[i]
            if len(curr) < min_words and i + 1 < len(tmp):
                merged = curr + tmp[i + 1]
                grouped.append(" ".join(merged))
                i += 2
            else:
                grouped.append(" ".join(curr))
                i += 1

    return grouped[:max_sentences]
```

```python
def generate_ocr_images(sentences, lang, font_path):
    out_dir = os.path.join(OUTPUT_BASE_DIR, lang)
    os.makedirs(out_dir, exist_ok=True)

    label_path = os.path.join(out_dir, "labels.csv")
    font = ImageFont.truetype(font_path, 32)

    def wrap_text_pixels(text, max_width):
        # Wrap text based on pixel width to avoid clipping for CJK and Latin text
        lines = []
        current = ""
        for ch in text:
            trial = current + ch
            bbox = font.getbbox(trial)
            width = bbox[2] - bbox[0]
            if width <= max_width:
                current = trial
            else:
                if current:
                    lines.append(current)
                current = ch
        if current:
            lines.append(current)
        return lines

    with open(label_path, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["filename", "language", "text"])

        max_width = 760  # leave horizontal padding
        padding = 20
        line_height = (font.getbbox("A")[3] - font.getbbox("A")[1]) + 8

        for i, text in enumerate(sentences):
            lines = wrap_text_pixels(text, max_width)
            if not lines:
                continue

            img_height = padding * 2 + line_height * len(lines)
            img_width = max_width + padding * 2

            img = Image.new("RGB", (img_width, img_height), "white")
            draw = ImageDraw.Draw(img)

            y = padding
            for line in lines:
                draw.text((padding, y), line, fill="black", font=font)
                y += line_height

            filename = f"{lang}_{i+1:03d}.png"
            img.save(os.path.join(out_dir, filename))
            writer.writerow([filename, lang, text])
```

```python
# Build 1-hour and 7.5-hour audio per language (en/ms/zh) by concatenating short clips
# Output: WAV 16 kHz, 16-bit PCM (optimized for audio ML models)
import os, subprocess, shutil, random
from tqdm import tqdm

DATASET_DIR = os.path.join(os.getcwd(), 'Dataset')
LANG_DIRS = {
    'en': os.path.join(DATASET_DIR, 'en', 'en'),
    'ms': os.path.join(DATASET_DIR, 'ms', 'ms'),
    'zh': os.path.join(DATASET_DIR, 'zh', 'zh'),
}
OUT_DIR = os.path.join(DATASET_DIR, 'compiled_audio')
OUT_DIR = os.path.join(DATASET_DIR, 'audio_active')
os.makedirs(OUT_DIR, exist_ok=True)

TARGET_SECONDS = 27000  # 7.5 hours
AUDIO_EXTS = {'.wav', '.mp3', '.flac', '.m4a', '.ogg'}


def has_ffmpeg():
    return shutil.which('ffmpeg') is not None and shutil.which('ffprobe') is not None


def list_audio_files(folder):
    if not os.path.isdir(folder):
        return []
    files = []
    for name in os.listdir(folder):
        path = os.path.join(folder, name)
        if os.path.isfile(path) and os.path.splitext(name)[1].lower() in AUDIO_EXTS:
            files.append(path)
    # Shuffle for variety; seed for reproducibility
    random.seed(0)
    random.shuffle(files)
    return files


def ffprobe_duration(path):
    # Returns float seconds or 0.0 on failure
    try:
        result = subprocess.run(
            ['ffprobe', '-v', 'error', '-show_entries', 'format=duration',
             '-of', 'default=noprint_wrappers=1:nokey=1', path],
            stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True, shell=False
        )
        return float(result.stdout.strip())
    except Exception:
        return 0.0


def select_until_duration(paths, target_seconds, desc=None):
    selected = []
    total = 0.0
    iterator = tqdm(paths, desc=desc) if desc else paths
    for p in iterator:
        d = ffprobe_duration(p)
        if d <= 0:
            continue
        selected.append(p)
        total += d
        if total >= target_seconds:
            break
    return selected, total


def write_concat_list(list_path, audio_paths):
    with open(list_path, 'w', encoding='utf-8') as f:
        for p in audio_paths:
            # Quote path; concat demuxer needs `file 'path'`
            f.write(f"file '{p}'\n")


def concat_to_wav(list_path, out_path):
    # Re-encode to WAV 16 kHz, 16-bit PCM for audio ML (wav2vec2, Conformer, etc.)
    cmd = ['ffmpeg', '-y', '-f', 'concat', '-safe', '0', '-i', list_path,
           '-c:a', 'pcm_s16le', '-ar', '16000', out_path]
    subprocess.run(cmd, check=True, shell=False)


if not has_ffmpeg():
    print("ffmpeg/ffprobe not found. Please install FFmpeg and ensure it's in PATH.")
    print("Windows quick install: download https://www.gyan.dev/ffmpeg/builds/ (release full), unzip, add bin to PATH.")
else:
    results = {}
    for lang, folder in LANG_DIRS.items():
        files = list_audio_files(folder)
        if not files:
            print(f"[{lang}] No audio files found in {folder}")
            continue
        selected, total = select_until_duration(files, TARGET_SECONDS, desc=f"{lang} selecting")
        if not selected:
            print(f"[{lang}] Could not select any audio with valid duration.")
            continue
        list_file = os.path.join(OUT_DIR, f'{lang}_list.txt')
        out_file = os.path.join(OUT_DIR, f'{lang}_7.5h.wav')
        write_concat_list(list_file, selected)
        try:
            concat_to_wav(list_file, out_file)
            results[lang] = {
                'clips': len(selected),
                'seconds': round(total, 2),
                'output': out_file
            }
            print(f"[{lang}] Built {out_file} using {len(selected)} clips (~{round(total/60,1)} min)")
        except subprocess.CalledProcessError as e:
            print(f"[{lang}] ffmpeg concat failed: {e}")
    if results:
        print("\n✅ Summary (WAV 16 kHz, 16-bit PCM - ready for audio ML models):")
        for lang, info in results.items():
            print(f" - {lang}: {info['clips']} clips, {info['seconds']}s -> {info['output']}")
```

```python
# Install CUDA-enabled PyTorch for GPU support
import subprocess
import sys

print("🔧 Installing CUDA-enabled PyTorch...\n")
print("⏳ This may take 5-10 minutes for large packages...\n")

# Uninstall CPU version and install CUDA version
commands = [
    [sys.executable, "-m", "pip", "uninstall", "torch", "torchvision", "torchaudio", "-y"],
    [sys.executable, "-m", "pip", "install", "torch", "torchvision", "torchaudio", 
     "--index-url", "https://download.pytorch.org/whl/cu118"]
]

for i, cmd in enumerate(commands, 1):
    print(f"Step {i}/{len(commands)}: {' '.join(cmd[-3:])}")
    result = subprocess.run(cmd, capture_output=True, text=True)
    if result.returncode == 0:
        print(f"   ✅ Success\n")
    else:
        print(f"   ⚠️ Warning: {result.stderr[:200]}\n")

print("✅ Installation complete! Please restart the kernel and run the GPU check cell again.")
```
